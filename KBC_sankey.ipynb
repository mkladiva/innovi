```python
data=pd.read_csv('kbc_group_statement.csv')
```

```python
x=data.loc[:,'line':'Group']
x = x[~x['level'].isna()]
x['parent line']=x['parent line'].astype(pd.Int32Dtype())
x['Group']=pd.to_numeric(x['Group'].str.replace(',',''))
x['pos']=pd.to_numeric(x['pos'])
x = x.sort_values('level')
x = x.reset_index(drop=True)
x = x.reset_index().set_index('line')
x
```

```python
import numpy as np
```

```python
labels = []
ys = []
xs = []
sources = []
targets = []
values = []
colors = []
for k, r in x.sort_values('level').iterrows():    
    label = r.iloc[4]
    labels.append(label)
    ys.append(r['level'])
    xs.append(r['pos'])
    value = r['Group']
    parent = r['parent line']
    if not pd.isna(parent):        
        if (value>10) and ('RESULT' not in label):
            sources.append(r['index'])
            targets.append(x.loc[parent,'index'])
            values.append(value)
            colors.append('blue')
        elif value<-10:
            sources.append(x.loc[parent,'index'])
            targets.append(r['index'])
            values.append(-value)
            colors.append('red')
        elif 'RESULT' in label:
            sources.append(x.loc[parent,'index'])
            targets.append(r['index'])
            values.append(value)
            colors.append('gold')
            

xs = np.array(xs)
xs = xs/3000
ys = np.array(ys)
ys = ys/7
fig = go.Figure(go.Sankey(
    node = {
        "label": labels,
        "x": ys,
        "y": xs,
        'pad':10},  # 10 Pixels
    link = {
        "source": sources,
        "target": targets,
        "value": values,
        "color": colors})
               )

fig.show()        
        
```
